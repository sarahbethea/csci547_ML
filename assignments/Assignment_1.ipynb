3. Show that the eigenvectors are orthogonal to one another (e.g. their inner product is zero). This is true for eigenvectors from real, symmetric matrices. Typically we use the orthogonal basis of our standard x, y, and z, Cartesian coordinates, which allows us, if we combine them linearly, to represent any point in a 3D space.
```python
# --- Question 1. ---
A = np.array([[1, 2, 3], [2, 4, 5], [3, 5, 6]])

# Because A is a symmetrical matrix, we can use numpy.linalg.eigh function
eigenvalues, eigenvectors = np.linalg.eigh(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

lambda1 = eigenvalues[0]
v1 = eigenvectors[:, 0]
print("Eigenvector 1:", v1, "Eigenvalue:", lambda1)

lambda2 = eigenvalues[1]
v2 = eigenvectors[:, 1]
print("Eigenvector 2:", v2, "Eigenvalue:", lambda2)

lambda3 = eigenvalues[2]
v3 = eigenvectors[:, 2]
print("Eigenvector 3:", v3, "Eigenvalue:", lambda3)
```

```python
# There are multiple rows for each county in the radon dataset, one for each test period. Keep only most recent period.

radon = radon_raw[
    (radon_raw["Start Year"] == 2008) &
    (radon_raw["End Year"] == 2017)
].copy()

print("Cancer data shape:", cancer_raw.shape)
print("Radon data shape:", radon.shape)

display(cancer_raw.head())
display(radon.head())
```

```python
# Check for null values in cancer rate column
# To do so, convert this column to numeric so that pandas can identify as null any row with non-numeric entry

rate_column = (
    "Age-Adjusted Incidence Rate([rate note]) - cases per 100,000"
)

numeric_cancer_rates = pd.to_numeric(
    cancer_raw[rate_column],
    errors="coerce"
)

print(numeric_cancer_rates.isna().sum())

# There are 381 non-numeric rows out of 3179 total rows 
```

```python
# Check for null values in mean radon level value column 
# To do so, convert this column to numeric so that pandas can identify as null any row with non-numeric entry

numeric_radon_levels = pd.to_numeric(
    radon["Value"],
    errors="coerce"
)

print(numeric_radon_levels.isna().sum())

# There are 269 non-numeric rows out of 2930 total rows 
```

```python
# Create dataframes for cancer rate and radon level data and remove non-numeric rows 
cancer = cancer_raw.copy()

cancer["cancer_rate"] = numeric_cancer_rates 
radon["radon_level"] = numeric_radon_levels

cancer = cancer.dropna(subset=["cancer_rate"])
radon = radon.dropna(subset=["radon_level"])

print("Cleaned cancer shape:", cancer.shape)
print("Cleaned radon shape:", radon.shape)

print("Missing cancer rates:", cancer["cancer_rate"].isna().sum())
print("Missing radon levels:", radon["radon_level"].isna().sum())
```

```python
# Remove first row of cancer dataframe, which provides summary nationwide data
cancer = cancer[cancer["FIPS"] != 0]

display(cancer.head())
```

```python
# Ensure FIPS column for both data sets uses same format
cancer["FIPS"] = (
    cancer["FIPS"]
    .astype(int)
    .astype(str)
    .str.zfill(5)
)

radon["FIPS"] = (
    radon["CountyFIPS"]
    .astype(int)
    .astype(str)
    .str.zfill(5)
)

# Confirm each FIPS code occurs only once
print("Duplicate cancer FIPS:", cancer["FIPS"].duplicated().sum())
print("Duplicate radon FIPS:", radon["FIPS"].duplicated().sum())
```

```python
# Select only necessary columns, format column names, and merge into single dataframe
cancer_for_merge = cancer[
    ["FIPS", "County", "cancer_rate"]
].rename(columns={"County": "county"})

radon_for_merge = radon[
    ["FIPS", "State", "radon_level"]
].rename(columns={"State": "state"})


# Using how="inner" here indicates inner join, which keeps only counties that have data in both datasets
combined = pd.merge(
    cancer_for_merge,
    radon_for_merge,
    on="FIPS",
    how="inner",
    validate="one_to_one"
)

print("Cancer rows:", len(cancer_for_merge))
print("Radon rows:", len(radon_for_merge))
print("Merged rows:", len(combined))

display(combined.head())
```

### Plots

```python
# Cancer rate histogram

plt.figure(figsize=(8, 5))

plt.hist(
    combined["cancer_rate"],
    bins=30,
    color="steelblue",
    edgecolor="black"
)

plt.title("Distribution of County Lung Cancer Rates")
plt.xlabel("Age-Adjusted Incidence Rate per 100,000")
plt.ylabel("Number of Counties")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    combined["radon_level"],
    bins=30,
    color="orange",
    edgecolor="black"
)

plt.axvline(
    4,
    color="red",
    linestyle="--",
    label="4 pCi/L threshold"
)

plt.title("Distribution of County Radon Levels")
plt.xlabel("Average Radon Level (pCi/L)")
plt.ylabel("Number of Counties")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
# Calculate correlation, draw best-fit line, create scatter plot
correlation = combined["radon_level"].corr(
    combined["cancer_rate"]
)

print(f"Correlation: {correlation:.4f}")

x = combined["radon_level"]
y = combined["cancer_rate"]

slope, intercept = np.polyfit(x, y, 1)

x_line = np.linspace(x.min(), x.max(), 100)
y_line = slope * x_line + intercept

plt.figure(figsize=(8, 5))

plt.scatter(
    x,
    y,
    alpha=0.3,
    s=20,
    color="steelblue"
)

plt.plot(
    x_line,
    y_line,
    color="red",
    linewidth=2,
    label="Regression line"
)

plt.title(
    f"County Radon Levels and Lung Cancer Rates\n"
    f"Correlation = {correlation:.3f}"
)

plt.xlabel("Average Radon Level (pCi/L)")
plt.ylabel("Age-Adjusted Cancer Rate per 100,000")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Cancer rates below and above 4 pCi/L

below_threshold = combined.loc[
    combined["radon_level"] < 4,
    "cancer_rate"
]

above_threshold = combined.loc[
    combined["radon_level"] >= 4,
    "cancer_rate"
]

plt.figure(figsize=(8, 5))

plt.boxplot(
    [below_threshold, above_threshold],
    patch_artist=True,
    showmeans=True
)

plt.xticks(
    [1, 2],
    ["Below 4 pCi/L", "At or Above 4 pCi/L"]
)

plt.title("County Lung Cancer Rates by Radon Category")
plt.xlabel("Radon Category")
plt.ylabel("Age-Adjusted Cancer Rate per 100,000")
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

print("Below 4 pCi/L:")
print(below_threshold.describe())

print("\nAt or above 4 pCi/L:")
print(above_threshold.describe())
```

### Results
Counties with radon levels below 4 pCi/L had an average age-adjusted lung cancer incidence rate of approximately 63.3 cases per 100,000, compared with 60.7 cases per 100,000 among counties at or above 4 pCi/L. The median cancer rate was also slightly higher in the lower-radon group. However, both groups had similar variability and their distributions overlapped substantially. These results do not show the expected pattern of higher county cancer rates above the 4 pCi/L threshold. If anything, the descriptive statistics show a small difference in the opposite direction.
