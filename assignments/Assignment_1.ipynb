```python
import numpy as np

A = np.array([[1, 2, 3], [2, 4, 5], [3, 5, 6]])
b = np.array([[-1], [3], [8]])
c = np.array([[4], [-3], [6]])
I = np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]])


print("\n1.", np.matmul(A, A))
print("\n2.", np.matmul(A, A.T))
print("\n3.", np.matmul(A, b))
# 4. Cannot be computed. A has dimensions (3, 3), b^T has dimensions (1, 3). The inner dimensions do not match. 
# 5. Cannot be computed. b has dimensions (3, 1), A has dimensions (3, 3). Inner dimensions do not match. 
print("\n6.", np.matmul(b.T, A))
# 7. Cannot be computed. b has dimensions (3, 1). (3, 1) * (3, 1) does not have matching inner dimensions. 
print("\n8.", np.matmul(b.T, b))
print("\n9.", np.matmul(b, b.T))
# 10. Cannot be computed. For matrix addition, matrices must have the same dimensions. b has dimensions (3, 1). c^T has dimensions (1, 3)
# 11. Cannot be computed. b^T has dimensions (1, 3). (1, 3) * (1, 3) does not have matching inner dimensions. 
print("\n12.", np.matmul(np.linalg.matrix_power(A, -1), b))
print("\n13.", np.multiply(A, A))
print("\n14.", np.multiply(b, c))
```

1. Calculate the eigenvalues and corresponding eigenvectors of matrix $\mathbf{A}$ above, from the last question.
2. Choose one of the eigenvector/eigenvalue pairs, $\mathbf{v}$ and $\lambda$, and show that $\mathbf{A} \mathbf{v} = \lambda \mathbf{v}$. This relationship extends to higher orders: $\mathbf{A} \mathbf{A} \mathbf{v} = \lambda^2 \mathbf{v}$
3. Show that the eigenvectors are orthogonal to one another (e.g. their inner product is zero). This is true for eigenvectors from real, symmetric matrices. In three dimensions or less, this means that the eigenvectors are perpendicular to each other. Typically we use the orthogonal basis of our standard x, y, and z, Cartesian coordinates, which allows us, if we combine them linearly, to represent any point in a 3D space. But any three orthogonal vectors can do the same. We will see this property is used in PCA to identify the dimensions of greatest variation in our data when we discuss dimensionality reduction.

**ANSWER:**

```python
# --- Question 1. ---
A = np.array([[1, 2, 3], [2, 4, 5], [3, 5, 6]])

# Because A is a symmetrical matrix, we can use numpy.linalg.eigh function
eigenvalues, eigenvectors = np.linalg.eigh(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

lambda1 = eigenvalues[0]
v1 = eigenvectors[:, 0]
print("Eigenvector 1:", v1, "Eigenvalue:", lambda1)

lambda2 = eigenvalues[1]
v2 = eigenvectors[:, 1]
print("Eigenvector 2:", v2, "Eigenvalue:", lambda2)

lambda3 = eigenvalues[2]
v3 = eigenvectors[:, 2]
print("Eigenvector 3:", v3, "Eigenvalue:", lambda3)
```

```python
# --- Question 2 ---

# Show that Av = lambda * v 
Av = A @ v3
lambda_v = v3 * lambda3
print(Av)
print(lambda_v)
print(np.allclose(Av, lambda_v))

# Show that it applies to higher order equation as well
# AAv = lambda^2 * v
AAv = A @ A @ v3
lambda_squared_v = (lambda3**2) * v3
print(AAv)
print(lambda_squared_v)
print(np.allclose(AAv, lambda_squared_v))
```

```python
# --- Question 3 ---
# Two vectors are orthogonal when their dot product is zero. 
# The nonzero here are extremely small values, basically zero
print("v1 dot v2:", np.dot(v1, v2))
print("v1 dot v3:", np.dot(v1, v3))
print("v2 dot v3:", np.dot(v2, v3))
```

```python
# --- PART 1 ---
# loc=2: mean=2
# scale=1: standard deviation=1
# size=10**4: generate 10^4 observations

rng = np.random.default_rng(42)

X = rng.normal(
    loc=2,
    scale=1,
    size=10**4
)

# --- PART 2 ---
print("std:", np.std(X))
print("mean:", np.mean(X))
```

```python
# --- PART 3 ---
plt.figure(figsize=(8, 5))
plt.hist(X, bins=30, edgecolor='black')
plt.title('Distribution of X', fontsize=14)
plt.xlabel('Values', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
# --- PART 4 ---
print("90th percentile:", np.percentile(X, 90))

# --- PART 5 ---
print("99th percentile:", np.percentile(X, 99))
```
